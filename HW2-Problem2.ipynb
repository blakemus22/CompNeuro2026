```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def bisection(a,b, tol, func):
    lower = a
    upper = b

    while upper - lower > tol:
        mid = (lower+upper)/2
        f_lower = func(lower)
        f_upper = func(upper)
        f_mid = func(mid)
        
        if f_mid*f_lower > 0:
            lower = mid
            print("New interval:", lower, "to", upper)
        else:
            upper = mid
            print("New interval:", lower, "to", upper)

    # once we are here, the difference between lower and upper is less than the tolerance
    # we will return the midpoint

    return (lower + upper) / 2
```

```python
col_list = ["pink", "mediumorchid", "navy"]
mg = [0, .1, 1]
v_0 = 16.13

for mg_val in range(3):
    v = np.linspace(-90, 0, 500)
    i_n = 1/(1+mg[mg_val]/3.57 * np.e**(-v/v_0)) * v
    
    plt.plot(v, i_n, color = col_list[mg_val], label = "Voltage with Mg_2+ = " + str(mg[mg_val]))
plt.title("Voltage plots with different magnesium concentrations")
plt.legend()
plt.xlabel("Voltage(mV)")
plt.ylabel("I")
sns.despine()
plt.grid(alpha = .3)
plt.savefig("mg_concentration_plot", dpi=300, bbox_inches="tight")
plt.show()
```
