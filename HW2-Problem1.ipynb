```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a_vals = np.linspace(0.01,.99,500)
```

```python
c_vals = (1-2*a_vals)/(np.sqrt(a_vals*(1-a_vals)))
```

```python
plt.plot(a_vals, c_vals, 
         color="#d62728", linewidth=1.2, label = "Active Front speed")
plt.axhline(y = 2, linestyle = "--", color = "grey", label = "Passive Apparent Speed")
plt.axhline(y = 0, color = "black")
plt.axvline(x = (2-np.sqrt(2))/4, linestyle = "--",
            label = "A = 0.14645")
plt.title("A and C plotted against each other", fontsize=13)
plt.xlabel("A", fontsize=11)
plt.ylabel("C", fontsize=11)
plt.legend(loc = "upper right", frameon = False)
plt.grid(alpha=0.3)
plt.savefig("a_and_c_plot", dpi=300, bbox_inches="tight")

plt.show()
```
